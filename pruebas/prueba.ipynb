```python
import folium
import json

# Carga los límites geográficos del estado deseado (por ejemplo, California)
with open('suarez_alrededores.geojson') as f:
    geo_json_data = json.load(f)

# Crea un objeto de mapa centrado en las coordenadas del estado
mapa = folium.Map()

# Agrega la capa GeoJSON al mapa y aplica un estilo
folium.GeoJson(
    geo_json_data,
    style_function=lambda feature: {
        'fillColor': 'green',  # Color de relleno del estado
        'color': 'black',      # Color del contorno del estado
        'weight': 2,           # Grosor del contorno del estado
        'fillOpacity': 0.5     # Opacidad del relleno del estado
    }
).add_to(mapa)

# Muestra el mapa
mapa
```
